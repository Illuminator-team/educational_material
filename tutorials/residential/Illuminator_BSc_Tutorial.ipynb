# 💡Plotting
We can visualise the results of your simulation by writing a function to plot them!

The function in the following block does just that!

```python
import pandas as pd
import matplotlib.pyplot as plt

#__________________________________________________________________
#
#  The function below plots some columns of the outputs.
#  We define it here so that we can easily use it later on!
#  Next time we want to plot our csv, we will only have to run:
#                   plot_csv('./out_CSV.csv')
#__________________________________________________________________

def plot_csv(csv_path):
    # Load CSV (handle BOM just in case)
    df_latest5 = pd.read_csv(csv_path, encoding='utf-8-sig')

    # Clean up column names
    df_latest5.columns = df_latest5.columns.str.strip()
    df_latest5.columns = df_latest5.columns.str.replace('\ufeff', '')  # remove BOM if present

    # OPTIONAL: Debug print
    # print(df_latest5.columns.tolist())

    # Parse the datetime column
    df_latest5['datetime'] = pd.to_datetime(df_latest5['date'], errors='coerce')

    # Drop rows with unparseable datetime
    df_latest5 = df_latest5.dropna(subset=['datetime'])

    # Set datetime as index
    df_latest5.set_index('datetime', inplace=True)

    # Corrected columns of interest (based on your actual CSV columns!)
    plot_columns_latest5 = {
        'Household Demand (kW)': 'Load1.load_dem',
        'PV Generation (kW)': 'PV1.pv_gen',
        'Battery SOC (%)': 'Battery1.soc',
        'Battery Power Out (kW)': 'Battery1.p_out',
        'EV Demand (kW)': 'EV1.demand',
        'Residual Load (kw)': 'Controller1.res_load'
    }

    # Prepare dataframe for plotting
    df_plot_latest5 = df_latest5[list(plot_columns_latest5.values())].rename(columns=plot_columns_latest5)

    # Plot
    plt.figure(figsize=(15, 10))

    for i, (label, col) in enumerate(plot_columns_latest5.items(), 1):
        plt.subplot(len(plot_columns_latest5), 1, i)

        if label == 'Residual Load (kw)':
            # Residual Demand = - Residual Load
            residual_load = -df_plot_latest5[col]  # Column is actually residual demand, so we negate it

            # Plot line in yellow/orange
            plt.plot(df_plot_latest5.index, residual_load, color='orange', linewidth=2, label='Residual Load')

            # Fill above zero
            plt.fill_between(df_plot_latest5.index, residual_load, 0,
                             where=(residual_load > 0),
                             facecolor='green', alpha=0.5, interpolate=True, label='Excess Energy')

            # Fill below zero
            plt.fill_between(df_plot_latest5.index, residual_load, 0,
                             where=(residual_load <= 0),
                             facecolor='red', alpha=0.5, interpolate=True, label='Energy from Grid')
        else:
            # All other lines: yellowish/orange
            plt.plot(df_plot_latest5.index, df_plot_latest5[col], color='orange', linewidth=2, label=label)

        plt.ylabel(label)
        plt.legend()
        plt.grid(True)

    plt.tight_layout()
    plt.show()
```

Every time we wish to use the above function we can simply run the following command:

```python
plot_csv('./out_CSV.csv')
```

# 💡 Changing the simulation parameters

The Illuminator simulator offers some functions as well, since we have installed it in this environment we can use them by calling them.

A useful function in our case is the `set_scenario_param(,)` function.

We call this function for a specific scenario and give it the name of the parameter we wish to edit along with its new value.

The code block below does this with the start_time and end_time parameters of the scenario, thus changing the day of the simulation.

```python
# I am a romantic researcher, I want to simulate valentine's day

simulation_BSc.set_scenario_param('start_time', '2012-14-02 00:00:00')
simulation_BSc.set_scenario_param('end_time', '2012-14-02 23:45:00')

# Note that the accepted format is YYYY-DD-MM HH:MM:SS
# Note that each simulation time step is 15 minutes by default.
```

## 📝Sizing of Components
Look at the graphs we have printed so far. Note the battery SOC.

We can edit our simulation parameters to change the sizing of the simulated components, specifically, let's play around with PV and the Battery.

This time around, we will be using the function `set_model_param(model_name= , parameter= , value= )`.

Look at the example below, change the capacity of the PV and the battery for your assigned day. Then run the simulation and examine your plots.

```python
CONFIG_FILE = 'Tutorial4.yaml'
simulation_BSc = Simulation(CONFIG_FILE)

simulation_BSc.set_model_param(model_name='PV1', parameter='cap', value= 0) # < change this
simulation_BSc.set_model_param(model_name='Battery1', parameter='max_energy', value= 0) # change this


# run the simulation
simulation_BSc.run()
```
