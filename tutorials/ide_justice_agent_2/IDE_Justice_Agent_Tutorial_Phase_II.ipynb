## 💡New Concept

The idea of sufficiency introduces a new way of understanding justice within the energy hub. Until this point, the system primarily evaluated whether energy flows remained balanced, whether shortages or surpluses existed, whether reciprocity was maintained, and whether households contributed fairly to collective stability.

However, balance alone cannot distinguish between essential consumption, comfortable consumption,

and excessive consumption. A community may therefore remain technically balanced while still reproducing forms of ecological excess or social inequality.

Sufficiency therefore asks “How much energy is enough for a good life?” rather than “How much energy can we produce or consume?”

Research has shown that human well-being does not require unlimited energy consumption. Instead, there exists a range of consumption within which people can live safely, comfortably, and with dignity. This range is referred to as the sufficiency corridor.

The corridor defines a lower boundary below which deprivation begins, and an upper boundary beyond which consumption becomes excessive. Rather than maximizing energy use, sufficiency aims to maintain energy use within this socially and ecologically acceptable range. The sufficiency corridor therefore does not simply measure energy demand. It defines what forms of living are considered acceptable, necessary or excessive within the community.  

The lower boundary of the corridor is the sufficiency floor. This represents the minimum level of energy required to support a decent standard of living.

This does not refer merely to biological survival. Theoretical work on decent living standards argues that human well-being depends on a broader set of material and social conditions. These include thermal comfort, food preparation, lighting, communication, mobility, hygiene, healthcare access, and even participation in social life. The floor therefore represents the minimum energy necessary to enable dignified participation in society.

Importantly, the floor is normative, contextual, and socially constructed. It is not a universal fixed number. Thus, what counts as “enough” energy depends on climate, household composition, dwelling conditions, cultural expectations, technological systems, and local practices.

For example:

* a poorly insulated dwelling requires more heating energy than an efficient dwelling,

* colder climates require higher thermal needs,

* larger households benefit from sharing efficiencies,

* and social participation today may require internet access or communication technologies that were previously unnecessary.

The sufficiency floor therefore reflects the minimum conditions required for meaningful social and physical well-being under specific societal conditions.

The upper boundary of the corridor is the sufficiency ceiling. This represents the point beyond which additional energy consumption no longer meaningfully contributes to well-being and instead becomes socially or ecologically excessive.

Theoretical work on sufficiency repeatedly shows that increases in energy consumption eventually produce diminishing returns for human well-being. After a certain threshold, additional heating, appliance ownership, private mobility or luxury consumption, actually contribute relatively little to quality of life. Instead, this significantly increases resource use, infrastructure pressure and ecological impacts.

The ceiling therefore introduces an ethical limit to consumption. This is important because energy justice is not only concerned with deprivation. It is also concerned with overconsumption, unequal access and ecological sustainability.

In this sense, sufficiency differs fundamentally from efficiency. While efficiency asks: “How can we use less energy for the same outcome?”, sufficiency asks: “How much consumption is actually necessary in the first place?”

Like we observed, a system may become highly efficient while still enabling unsustainable levels of consumption. Sufficiency therefore places ethical limits on demand itself.

## 🧠 Reflection

The major challenge of sufficiency is therefore determining what level of energy constitutes “enough.” The literature consistently argues that this cannot be solved through purely technical optimization. Instead, sufficiency combines material requirements, social expectations, environmental constraints, and normative judgments.

To translate sufficiency into a measurable system condition, the model operationalizes the floor through household-specific energy requirements.  

Floor_h = {(Baseline energy per person × Occupants × Dwelling Size Factor) × Insulation Factor × Heating Factor × Appliance Factor × Seasonal Factor × Cultural Factor} + Mandatory Energy Needs

The ceiling is then derived as a socially negotiated threshold above the floor. The tolerance margin reflects communal norms, ecological priorities, and political decisions about acceptable excess.

Ceiling_h = Floor_h × (1 + tolerance margin)

## 🧠 Thought Experiment

A key assumption in the model is that everyone in the homogenous hub has the same capabilities. Under these conditions, households are generally capable of adapting behaviour, responding to incentives, and remaining within the sufficiency corridor. However, this assumption rarely reflects real energy systems. In reality, households do not possess the same ability to convert energy into well-being.

Capability theory begins from a critique of resource-based understandings of justice. It argues that justice cannot be evaluated simply through equal distribution of resources, infrastructure or rights because people do not possess the same ability to convert those resources into well-being.

The capability approach therefore shifts attention away from resources themselves and toward what people are actually able to do and be with those resources. Our homogenous model treats all households the same, assuming the same ability to respond to system signals and thus the same capability. However, capability theory suggests that this assumption hides important forms of injustice.  

This fundamentally complicates the sufficiency corridor. The sufficiency floor was previously treated as a shared minimum necessary for dignified living, while the ceiling represented a collectively negotiated limit beyond excessive consumption. But capability theory reveals that a universal corridor may not affect all households equally. The same floor may be insufficient for some households. The same ceiling may be restrictive for others. And the same behavioural request may produce radically different burdens across the community.

Two households may receive the same energy allocation yet experience completely different outcomes. One household may comfortably reduce heating demand during winter while another may require higher indoor temperatures because of illness, disability, age or care responsibilities. Similarly, one household may easily shift appliance usage across time while another may rely on constant energy access for medical equipment, remote work, childcare or mobility support.

The issue therefore is not only “How much energy does a household receive?” but rather “What can this household actually achieve with that energy?”

Capability theory refers to these differences as conversion factors: the personal, social and environmental conditions that shape how effectively households can convert energy into well-being. In the model, capability influences how households respond to energy reduction requests, how difficult behavioural adaptation becomes, whether sufficiency thresholds are realistically achievable, and whether the same intervention produces unequal burdens across households.

Importantly, capability is not treated as static. A household may become low capability due to illness, aging, extreme weather, infrastructure failure, changing family conditions,or unexpected energy dependence. This means vulnerability is not external to the energy system. It can emerge from within the system itself over time.

The model thus begins asking:

* Balanced for whom?

* Who is able to remain sufficient under these conditions?

* Whose needs fit comfortably within the corridor?

* Which households become vulnerable when collective limits are imposed?

# 💡New Concept

The introduction of differentiated capability reveals that sufficiency is not only a technical problem of optimization. So sufficiency governance cannot be reduced to distributive justice alone. Justice also depends on recognition, participation and representation. It is also a political problem about whose needs are recognized, whose vulnerabilities are visible and who participates in defining collective limits. This introduces two critical dimensions of justice: recognition and procedure.

**Recognition Justice**
> Recognition Justice looks at whose conditions, vulnerabilities, and ways of living are acknowledged within the system. The simulations show that some households are more visible than others or fit normative assumptions while equal thresholds can produce unequal burdens. Recognition justice therefore requires acknowledging differentiated vulnerabilities, questioning standardized assumptions and identifying whose needs remain invisible.

**Procedural Justice**
> Procedural Justice looks at who participates in defining the rules of governance. The simulations show that the sufficiency ceiling is not purely technical but reflects collective values, political priorities and social negotiation. Procedural justice therefore reframes the energy hub not simply as a management system, but as a political arena where definitions of “enough” are continuously negotiated.
